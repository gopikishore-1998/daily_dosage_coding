```python
##Find the second largest number in a list (without sorting)
def get_second_largest (num):#Single-Pass Greedy Algorithm.
    first_largest = float('-inf')
    second_largest = float('-inf')
    for n in num:
        if n > first_largest:
            second_largest = first_largest
            first_largest = n
        elif n > second_largest and n != first_largest:
            second_largest = n
    return second_largest if second_largest != float('inf') else None
list1 = [223,220,44,55,0,33,786,221,222]
print("2nd Largest number from given List is", get_second_largest(list1))
```
